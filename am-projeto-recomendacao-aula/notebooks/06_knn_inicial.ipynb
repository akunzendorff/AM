```python
musica_A = {"nome": "Música A", "danceability": 0.8, "energy": 0.6}
musica_B = {"nome": "Música B", "danceability": 0.3, "energy": 0.2}
```

```python
# Distância Euclidiana - passa por cima dos prédios

dx = musica_A["danceability"] - musica_B["danceability"]
dy = musica_A["energy"] - musica_B["energy"]

print(f"Passo 1 - Diferença em cada eixo")
print(f"Danceability: {dx}")
print(f"Energy: {dy}")

print("\n")

print(f"Passo 2 - Elevar ao quadrado (para tirar negativos)")
print(f"Danceability: {dx**2}")
print(f"Energy: {dy**2}")

print("\n")

print(f"Passo 3 - Somar")
print(f"{dx**2} + {dy**2} = {dx**2 + dy**2}")

print("\n")

print(f"Passo 4 - Raíz quadrada")
d_euclidiana = np.sqrt(dx**2 + dy**2)
print(f"Resultado: {d_euclidiana}")
```

```python
# Distância Manhattan - passa pelas ruas, por cada quadra

dx = musica_A["danceability"] - musica_B["danceability"]
dy = musica_A["energy"] - musica_B["energy"]

print(f"Passo 1 - Diferença em cada eixo")
print(f"Danceability: {dx}")
print(f"Energy: {dy}")

print("\n")

print(f"Passo 2 - Somar")
d_manhattan = abs(dx) + abs(dy)
print(f"{abs(dx)} + {abs(dy)} = {d_manhattan}")

print("\n")
```

```python
# Para rodar essa parte é necessário rodar todas as anteriores primeiro
print(f"Comparação Euclidiana x Manhattan")
print(f"Euclidiana (linha reta): {d_euclidiana:.4f}")
print(f"Manhattan (quadras): {d_manhattan:.4f}")

# O manhattan sempre dá um valor maior do que a euclidiana, já que a euclidiana é em linha reta e a manhattan enfrenta obstaculos
```
